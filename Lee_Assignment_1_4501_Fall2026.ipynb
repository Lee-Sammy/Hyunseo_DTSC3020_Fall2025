## 1.5. The Kernel
The kernel is a program that executes the code inside your notebook and outputs the results. In the top right of your window, you can see a circle that indicates the status of your kernel. If the circle is empty (⚪), the kernel is idle and ready to execute code. If the circle is filled in (⚫), the kernel is busy running some code.

Next to every code cell, you'll see some text that says `In [...]`. Before you run the cell, you'll see `In [ ]`. When the cell is running, you'll see `In [*]`. If you see an asterisk (\*) next to a cell that doesn't go away, it's likely that the code inside the cell is taking too long to run, and it might be a good time to interrupt the kernel (discussed below). When a cell is finished running, you'll see a number inside the brackets, like so: `In [1]`. The number corresponds to the order in which you run the cells; so, the first cell you run will show a 1 when it's finished running, the second will show a 2, and so on.

You may run into problems where your kernel is stuck for an excessive amount of time, your notebook is very slow and unresponsive, or your kernel loses its connection. If this happens, try the following steps:
1. At the top of your screen, click **Kernel**, then **Interrupt**.
2. If that doesn't help, click **Kernel**, then **Restart**. If you do this, you will have to run your code cells from the start of your notebook up until where you paused your work.

# 2. Basic Python Syntax

### 2.1. Variables & Numeric Types
In Python, **variables** act as containers for storing data values. You don't need to declare their type explicitly.

We will explore three main numeric types:
* **int**: Integers (whole numbers)
* **float**: Floating-point numbers (decimals)
* **complex**: Complex numbers (with real and imaginary parts)

```python
# Variables & Assignment
# Assign values using the '=' operator
x = 10
y = 3.14

# Numeric Types
# int: Whole numbers
count = 42

# float: Decimal numbers
price = 19.99

# complex: Numbers with an imaginary part (denoted by 'j')
comp_num = 2 + 3j

print("--- Numeric Types ---")
print("Integer:", count)
print("Float:", price)
print("Complex:", comp_num)
```

### 2.2. Strings & Booleans
**Strings** are sequences of characters used for text. **Booleans** represent one of two values: `True` or `False`.

```python
# Strings
# Can use single or double quotes
first_name = "Data"
last_name = 'Scientist'

# String Operations
# Concatenation: Joining strings together
full_name = first_name + " " + last_name

# Repetition
cheer = "Hip! " * 2 + "Hooray!"

print("--- Strings ---")
print("Full Name:", full_name)
print("Cheer:", cheer)

# Boolean Values
is_raining = False
is_sunny = True

print("\n--- Booleans ---")
print("Is it sunny?", is_sunny)
# Comparison operators return Booleans
print("Is 10 greater than 2?", 10 > 2)
```

### 2.3. Functions & Type Conversion
Python has **Built-in Functions** provided by the language, and allows you to create **User-defined Functions**.

You can also check the **type** of a variable and **convert** between types.

```python
# Built-in Functions
text = "Python"
# len() returns the length of the string
print("Length of 'Python':", len(text))

# Type Inspection
val = "50"
print("\nType of val:", type(val))

# Type Conversion (Casting)
# Convert string to integer
val_int = int(val)
print("Converted to int:", val_int, type(val_int))

# User-defined Functions
def greet_user(name):
    """This function prints a greeting."""
    return "Hello, " + name + "!"

print("\n--- User Function ---")
msg = greet_user("Alice")
print(msg)
```

### 2.4. Loops & Table Operations
**For Loops** let you iterate over sequences (like a string).

For **Table Operations**, we use the `pandas` library, creating a `DataFrame` (a table).

```python
# Simple For Loop over a String
print("--- Loop Output ---")
word = "Code"
for char in word:
    print(char)

# Table Operations using Pandas
import pandas as pd

# Create a dictionary of data
data = {
    'ID': [101, 102, 103],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 90, 78]
}

# Create DataFrame
df = pd.DataFrame(data)

print("\n--- Table (DataFrame) ---")
display(df)

# Select a column
print("\nScores column:")
print(df['Score'])
```
